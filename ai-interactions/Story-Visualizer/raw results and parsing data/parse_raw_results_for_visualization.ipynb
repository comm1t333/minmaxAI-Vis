```python
#provided by gemini

def transform_df_to_nested_json(df):
    # Define the columns that represent the extracted attributes
    attribute_cols = questions

    result = []

    # Group by character/dataset level metadata
    group_cols = ["dataset_index", "story_function", "name", "true_language", "true_gender", "true_goodness"]

    for keys, group in df.groupby(group_cols):
        dataset_index, story_function, name, true_language, true_gender, true_goodness = keys

        steps_list = []
        # Sort by step_number to ensure the steps are ordered correctly (1 through 9)
        for _, row in group.sort_values("step_number").iterrows():
            extracted_attributes = {col: row[col] for col in attribute_cols if col in df.columns}

            step_dict = {
                "step_number": int(row["step_number"]),
                "story_snapshot": row["story_snapshot"],
                "extracted_attributes": extracted_attributes
            }
            steps_list.append(step_dict)

        entry = {
            "dataset_index": int(dataset_index),
            "story_function": story_function,
            "character_name": name,
            "true_language": true_language,
            "true_gender": true_gender,
            "true_goodness": true_goodness,
            "steps": steps_list
        }
        result.append(entry)

    return result
```

```python
path_data_small = "/content/data_small-01.json"
path_data_big = "/content/data_big-01.json"
```

```python
dfbig = pd.read_json(path_data_big)
bigdata = transform_df_to_nested_json(dfbig)

# To save to a file:
with open('big_data.json', 'w', encoding='utf-8') as f:
    json.dump(bigdata, f, indent=2, ensure_ascii=False)
```
